```python
from scipy.stats import ttest_ind
import numpy as np

# Separate groups
ad_samples = data[data["Condition"] == "AD"].drop("Condition", axis=1)
control_samples = data[data["Condition"] == "Control"].drop("Condition", axis=1)

deg_results = []

for gene in ad_samples.columns:
    stat, pval = ttest_ind(ad_samples[gene], control_samples[gene])
    log2fc = np.mean(ad_samples[gene]) - np.mean(control_samples[gene])
    deg_results.append([gene, log2fc, pval])

deg_df = pd.DataFrame(deg_results, columns=["Gene", "log2FC", "p_value"])
deg_df["-log10(p_value)"] = -np.log10(deg_df["p_value"])

print(deg_df.head())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(deg_df["log2FC"], deg_df["-log10(p_value)"], alpha=0.5)

plt.axvline(1, color='red', linestyle='--')
plt.axvline(-1, color='red', linestyle='--')
plt.axhline(1.3, color='blue', linestyle='--')

plt.xlabel("log2 Fold Change")
plt.ylabel("-log10(p-value)")
plt.title("Volcano Plot of Differentially Expressed Genes")

plt.show()
```

```python
# Significant genes
sig_deg = deg_df[deg_df["p_value"] < 0.05]

top_up = sig_deg.sort_values("log2FC", ascending=False).head(10)
top_down = sig_deg.sort_values("log2FC").head(10)

print("Top 10 Upregulated Genes:")
print(top_up)

print("\nTop 10 Downregulated Genes:")
print(top_down)
```
